```python
import sys
from pathlib import Path
from decimal import Decimal

# Add root directory to path
root_dir = Path().resolve().parent
if str(root_dir) not in sys.path:
    sys.path.append(str(root_dir))

import pandas as pd
from src.tax_engine import UgandaPAYECalculator, UgandaNSSFCalculator

paye_calc = UgandaPAYECalculator()
nssf_calc = UgandaNSSFCalculator()
```

```python
# Simulate calculations across various salary tiers
salaries = [Decimal("200000"), Decimal("300000"), Decimal("400000"), Decimal("500000"), Decimal("1500000")]

data = []
for sal in salaries:
    nssf = nssf_calc.calculate_tax(sal)
    paye = paye_calc.calculate_tax(sal)
    net = sal - (nssf + paye)
    data.append({
        "Gross Income (UGX)": sal,
        "NSSF (5%)": nssf,
        "PAYE": paye,
        "Net Income (UGX)": net
    })

df_tax = pd.DataFrame(data)
display(df_tax)
```
